```python
import numpy as np
```

Gaussian line profile

parameters 

wav = wavelength , shape = (N,), linespace, angstrom

wav0 = central wavelength , shape(N,),array,angstrom

sigma = line width ,float, angstrom

amp = amplitude or depth , float , dimension less

 


```python
wav = np.linspace(4990,5010,1000)
wav0 = np.asarray(5000.0)
print(wav0.ndim)
wave0 = np.array([5000,5000.50,5010])
sigma = 1.0
amp = 1.0
def Gaussian_line(wav,wav0,sigma,amp):
    result = amp*(np.exp(-(wav - wav0)**2) / (2 * sigma**2))
    return result
```

```python
profile = Gaussian_line(wav,wav0,sigma,amp)
profile.shape
```

```python
def Gaussian_line(wav,wav0,sigma,amp):
    result = amp*(np.exp(-(wav[:,np.newaxis] - wav0[np.newaxis,:])**2) / (2 * sigma**2))
    return result
profile = Gaussian_line(wav,wave0,sigma,amp)
profile.shape
```
